```python

# ===================================================================
# 3. 下载模型到指定目录
# ===================================================================
import os
from huggingface_hub import snapshot_download

# --- 用户配置 ---
# 你要下载的模型
repo_id = "google/medgemma-27b-text-it"
# 你想保存到的 Drive 路径 (可以自定义)
save_directory = "/content/drive/MyDrive/huggingface_models/medgemma-27b-text-it"
# --- 用户配置结束 ---

# 确保目标文件夹存在
os.makedirs(save_directory, exist_ok=True)

print(f"\n🚀 开始下载模型 {repo_id} 到 {save_directory} ...")
print("这个过程可能需要一些时间，请耐心等待...")

try:
    snapshot_download(
        repo_id=repo_id,
        local_dir=save_directory,
        local_dir_use_symlinks=False # 直接下载文件，而不是链接
    )
    print("\n🎉 模型成功下载到你的 Google Drive！")

    # 验证文件
    print("\n以下是下载目录中的文件列表:")
    !ls -lh {save_directory}

except Exception as e:
    print(f"\n❌ 下载过程中出现错误: {e}")
    print("请检查：")
    print("1. 是否已在 Hugging Face 网站上同意 medgemma 的使用条款。")
    print("2. 你的 Hugging Face Token 是否正确且有读取权限。")
    print("3. 你的 Google Drive 剩余空间是否足够 (medgemma-4b-it 大约需要 8-10 GB)。")
```
